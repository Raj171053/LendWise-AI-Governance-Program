```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

print("Preprocessing tools imported successfully.")
```

```python

numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()

print("Number of numerical columns:", len(numeric_features))
print("Number of categorical columns:", len(categorical_features))

print("\nNumerical columns:")
print(numeric_features)

print("\nCategorical columns:")
print(categorical_features)
```

```python

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created.")
```

```python

# Preserve the original test-set row indexes.
age_test = X_test["age_years"]

# Create exploratory age groups.
age_group_test = pd.Series(
    age_test.map(
        lambda age: "Under 25" if age < 25 else "25 and older"
    ),
    index=X_test.index,
    name="age_group"
)

print("Age-group counts in the test set:")
display(age_group_test.value_counts())
```

```python

# Assemble actual labels, predictions and age groups.
audit_results = pd.DataFrame({
    "actual": y_test,
    "predicted": y_pred,
    "age_group": age_group_test
})

def calculate_group_metrics(group):
    tn, fp, fn, tp = confusion_matrix(
        group["actual"],
        group["predicted"],
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn) if (fp + tn) > 0 else float("nan")
    fnr = fn / (fn + tp) if (fn + tp) > 0 else float("nan")

    return pd.Series({
        "sample_size": len(group),
        "good_credit_count": (group["actual"] == 0).sum(),
        "bad_credit_count": (group["actual"] == 1).sum(),
        "false_positives": fp,
        "false_negatives": fn,
        "false_positive_rate": fpr,
        "false_negative_rate": fnr
    })

age_fairness_results = (
    audit_results
    .groupby("age_group")
    .apply(calculate_group_metrics, include_groups=False)
)

display(age_fairness_results.round(4))
```

```python

from statsmodels.stats.proportion import proportion_confint

def wilson_interval(errors, total):
    if total == 0:
        return (float("nan"), float("nan"))

    lower, upper = proportion_confint(
        count=errors,
        nobs=total,
        alpha=0.05,
        method="wilson"
    )

    return round(lower, 4), round(upper, 4)

print("Wilson confidence-interval function ready.")
```

```python

for group_name, group in audit_results.groupby("age_group"):
    tn, fp, fn, tp = confusion_matrix(
        group["actual"],
        group["predicted"],
        labels=[0, 1]
    ).ravel()

    fpr_ci = wilson_interval(fp, fp + tn)
    fnr_ci = wilson_interval(fn, fn + tp)

    print(f"\n{group_name}")
    print(f"FPR: {fp / (fp + tn):.4f}, 95% CI: {fpr_ci}")
    print(f"FNR: {fn / (fn + tp):.4f}, 95% CI: {fnr_ci}")
```

```python

from statsmodels.stats.proportion import confint_proportions_2indep

# Counts from the verified age-group confusion matrices.
# First group: under 25
fp_under25 = 4
good_under25 = 20

# Second group: age 25 and older
fp_older = 16
good_older = 120

fpr_under25 = fp_under25 / good_under25
fpr_older = fp_older / good_older

difference = fpr_under25 - fpr_older

ci_low, ci_high = confint_proportions_2indep(
    count1=fp_under25,
    nobs1=good_under25,
    count2=fp_older,
    nobs2=good_older,
    method="newcomb",
    compare="diff",
    alpha=0.05
)

print("Under-25 FPR:", round(fpr_under25, 4))
print("Age-25-and-older FPR:", round(fpr_older, 4))
print("Difference in FPR:", round(difference, 4))
print("Difference in percentage points:", round(difference * 100, 2))
print(
    "95% CI for difference:",
    (round(ci_low, 4), round(ci_high, 4))
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

def evaluate_model(name, y_true, predictions):
    tn, fp, fn, tp = confusion_matrix(
        y_true, predictions, labels=[0, 1]
    ).ravel()

    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision (bad credit)": precision_score(
            y_true, predictions, zero_division=0
        ),
        "Recall (bad credit)": recall_score(
            y_true, predictions, zero_division=0
        ),
        "F1 (bad credit)": f1_score(
            y_true, predictions, zero_division=0
        ),
        "FPR": fp / (fp + tn) if (fp + tn) else float("nan"),
        "FNR": fn / (fn + tp) if (fn + tp) else float("nan")
    }

comparison = pd.DataFrame([
    evaluate_model("Baseline — all features", y_test, y_pred),
    evaluate_model(
        "Without foreign_worker",
        y_test,
        y_pred_no_foreign
    )
])

display(comparison.round(4))
```

```python
# Compare group-level errors for both models
# Reuses existing test labels and predictions

def group_error_rates(y_true, y_predicted, groups, model_name):
    audit = pd.DataFrame({
        "actual": np.asarray(y_true),
        "predicted": np.asarray(y_predicted),
        "group": np.asarray(groups)
    })

    results = []

    for group_name, data in audit.groupby("group"):
        tn = ((data["actual"] == 0) & (data["predicted"] == 0)).sum()
        fp = ((data["actual"] == 0) & (data["predicted"] == 1)).sum()
        fn = ((data["actual"] == 1) & (data["predicted"] == 0)).sum()
        tp = ((data["actual"] == 1) & (data["predicted"] == 1)).sum()

        results.append({
            "Model": model_name,
            "Group": group_name,
            "N": len(data),
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp,
            "FPR": fp / (fp + tn) if (fp + tn) else np.nan,
            "FNR": fn / (fn + tp) if (fn + tp) else np.nan
        })

    return results


# Preserve the original category labels from the same held-out test rows
foreign_groups = X_test["foreign_worker"].reset_index(drop=True)

group_comparison = pd.DataFrame(
    group_error_rates(
        y_test, y_pred, foreign_groups, "Baseline"
    )
    +
    group_error_rates(
        y_test, y_pred_no_foreign, foreign_groups,
        "Without foreign_worker"
    )
)

display(group_comparison.round(4))
```
